# Simulate sample means


```python
# Edit these defaults for interactive use.
task_id = 1
output_dir = "results/local-test"
```

```python
from pathlib import Path

import numpy as np
import pandas as pd

if type(task_id) is not int or task_id not in range(1, 6):
    raise ValueError("task_id must be an integer from 1 to 5")

output_dir = Path(output_dir)
output = output_dir / f"task-{task_id:03d}.csv"
if output.exists():
    raise FileExistsError(f"Refusing to overwrite: {output}")
```

## Run the simulation


```python
rng = np.random.default_rng(task_id)
sample_sizes = [10, 30, 100, 300, 1000]
sample_size = sample_sizes[task_id - 1]

results = pd.DataFrame({
    "task_id": task_id,
    "repetition": np.arange(1, 1001),
    "sample_size": sample_size,
    "sample_mean": rng.normal(size=(1000, sample_size)).mean(axis=1),
})
assert len(results) == 1000
results.head()
```

## Save the results


```python
output_dir.mkdir(parents=True, exist_ok=True)
results.to_csv(output, index=False, mode="x")
print(f"Saved {len(results)} repetitions to {output}")
```
